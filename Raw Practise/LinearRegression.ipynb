```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# House sizes in hundreds of square feet
X = np.array([1, 2, 3, 4, 5], dtype=float)

# House prices in thousands of dollars
y = np.array([2, 4, 6, 8, 10], dtype=float)
```

```python
# Initial parameters
w = 3.0
b = 0.0
```

```python

# Prediction
y_pred = w * X + b

print(y_pred)
```

```python
# loss function
def loss(y, y_pred):
    return np.mean((y_pred - y) ** 2)

print(loss(y, y_pred))
```

```python
# Gradient Descent
def train_linear_regression(X, y, learning_rate=0.01, epochs=1000):
    n = len(X)

    w = 0.0
    b = 0.0
    losses = [] 

    for epoch in range(epochs):
        # 1. Predict
        y_pred = w * X + b

        # 2. Calculate errors
        error = y_pred - y

        # 3. Calculate gradients
        dw = (2 / n) * np.sum(X * error) # Calculate how loss changes with weight
        db = (2 / n) * np.sum(error) # Calculate how loss changes with bias

        # 4. Update parameters
        w = w - learning_rate * dw # Adjust the slope to reduce loss
        b = b - learning_rate * db # Adjust the intercept to reduce loss

        # 5. Record loss after updating
        losses.append(loss(y, w * X + b))

    return w, b, losses


w, b, losses = train_linear_regression(X, y)

print("Weight:", w)
print("Bias:", b)
print("Final loss:", losses[-1])
```

```python
# loss plot

plt.plot(losses)
plt.xlabel("Epoch")
plt.ylabel("MSE Loss")
plt.title("Linear Regression Training")
plt.grid(True)
plt.show()
```
